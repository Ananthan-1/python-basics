## NUMPY FUNDAMENTALS

```python
import numpy as np
```

```python
arr=np.array([1,2,3,4])
print(arr)
print(arr.dtype)
print(arr.shape)
```

```python
#2D ARRAY
m1=np.array([[1,2,3,4],[5,6,7,8]])
print(m1)
print(m1.shape)
```

```python
print(arr[0])
print(arr[-1])
print(m1[0,2])
print(m1[:,1])
```

```python
arr1=np.array([10,20,30,40])
print(arr+arr1)
print(arr*2)
print(arr1/10)
```

```python
#NUMPY FUNCTIONS
print(np.mean(arr))
print(np.min(m1))
print(np.max(arr1))
print(np.sum(arr))
print(np.std(arr))
```

```python
print(np.zeros((3,3)))
print(np.ones((2,4)))
print(np.arange(0,20,2))
print(np.linspace(0,1,5))
```

```python
print(np.random.seed(0))
print(np.random.randint(1,100,size=5))
print(np.random.randn(3,3))
```
