## Lists, Dictionaries, and File Handling (Python)


```python
num=[10,20,30,40,50]
print(num)
print(num[0])
print(num[-1])
print(num[1:3])

num.append(60)
num.insert(2,25)
num.remove(50)
num
```

```python
#Square of even numbers in a list in range 0-18
sqlist=[x*x for x in range(19) if x%2==0]
print(sqlist)
```

```python
#Dictionary Operations
dict1={
    "Name":"Ananthan",
    "Age":23,
    "State":"Kerala"}
print(dict1["Name"])
print(dict1.get("cgp","Null"))
dict1["cgp"]=8.9
for key,value in dict1.items():
    print(key,"->",value)
```

```python
scores={
    "Maths":90,
    "Physics":88,
    "Python":89}
inverted={}
for subject,score in scores.items():
    inverted.setdefault(score,[]).append(subject)
print(inverted)
```
